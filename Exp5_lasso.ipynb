```python
# Step 1: Import required libraries and load dataset

import numpy as np
import pandas as pd

from tensorflow.keras.datasets import boston_housing

# Load dataset
(X_train, y_train), (X_test, y_test) = boston_housing.load_data()

# Check shapes
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
# Step 2: Feature Scaling

from sklearn.preprocessing import StandardScaler

# Initialize scaler
scaler = StandardScaler()

# Fit on training data and transform both train & test
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Verify scaling (mean ~0, std ~1)
print("Mean of scaled training data:", X_train_scaled.mean())
print("Std of scaled training data:", X_train_scaled.std())
```

```python
# Step 3: Linear Regression

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Initialize model
lr = LinearRegression()

# Train model
lr.fit(X_train_scaled, y_train)

# Predictions
y_pred_lr = lr.predict(X_test_scaled)

# Evaluation
mse_lr = mean_squared_error(y_test, y_pred_lr)
r2_lr = r2_score(y_test, y_pred_lr)

print("Linear Regression MSE:", mse_lr)
print("Linear Regression R2 Score:", r2_lr)
```

```python
# Step X: Multiple Linear Regression (using selected multiple features)

# Select a subset of features (example: first 5 features)
X_train_multi = X_train_scaled[:, :5]
X_test_multi = X_test_scaled[:, :5]

# Initialize model
mlr = LinearRegression()

# Train model
mlr.fit(X_train_multi, y_train)

# Predictions
y_pred_mlr = mlr.predict(X_test_multi)

# Evaluation
mse_mlr = mean_squared_error(y_test, y_pred_mlr)
r2_mlr = r2_score(y_test, y_pred_mlr)

print("Multiple Linear Regression MSE:", mse_mlr)
print("Multiple Linear Regression R2 Score:", r2_mlr)
```

```python
# Step 4: Lasso Regression

from sklearn.linear_model import Lasso

# Initialize model (alpha = regularization strength)
lasso = Lasso(alpha=0.1)

# Train model
lasso.fit(X_train_scaled, y_train)

# Predictions
y_pred_lasso = lasso.predict(X_test_scaled)

# Evaluation
mse_lasso = mean_squared_error(y_test, y_pred_lasso)
r2_lasso = r2_score(y_test, y_pred_lasso)

print("Lasso MSE:", mse_lasso)
print("Lasso R2 Score:", r2_lasso)
```

```python
# Step 5: Ridge Regression

from sklearn.linear_model import Ridge

# Initialize model
ridge = Ridge(alpha=1.0)

# Train model
ridge.fit(X_train_scaled, y_train)

# Predictions
y_pred_ridge = ridge.predict(X_test_scaled)

# Evaluation
mse_ridge = mean_squared_error(y_test, y_pred_ridge)
r2_ridge = r2_score(y_test, y_pred_ridge)

print("Ridge MSE:", mse_ridge)
print("Ridge R2 Score:", r2_ridge)
```

```python
# Step 6: Elastic Net

from sklearn.linear_model import ElasticNet

# Initialize model
elastic = ElasticNet(alpha=0.1, l1_ratio=0.5)

# Train model
elastic.fit(X_train_scaled, y_train)

# Predictions
y_pred_elastic = elastic.predict(X_test_scaled)

# Evaluation
mse_elastic = mean_squared_error(y_test, y_pred_elastic)
r2_elastic = r2_score(y_test, y_pred_elastic)

print("Elastic Net MSE:", mse_elastic)
print("Elastic Net R2 Score:", r2_elastic)
```

```python
# Coefficients comparison

print("Linear Regression Coefficients:\n", lr.coef_)
print("\nLasso Coefficients:\n", lasso.coef_)
print("\nRidge Coefficients:\n", ridge.coef_)
print("\nElastic Net Coefficients:\n", elastic.coef_)
```

```python
# Step 7: Comparison Table (Lasso, Ridge, Elastic Net)

comparison = pd.DataFrame({
    "Model": ["Lasso", "Ridge", "Elastic Net"],
    "MSE": [mse_lasso, mse_ridge, mse_elastic],
    "R2 Score": [r2_lasso, r2_ridge, r2_elastic]
})

print(comparison)
```
